```python
import pandas as pd
pd.set_option('display.width', 1000)  #

df = pd.read_csv("../data/feed-views.log", sep = "\t", header=None, names=['datetime', 'user'])
print(df)
df.info()
```

```python
df['datetime'] = df['datetime'].astype('datetime64[ns]')
print(df)
df.info()
```

```python
datetime_ser = df['datetime']
df['year'] = datetime_ser.dt.year
df['month'] = datetime_ser.dt.month
df['day'] = datetime_ser.dt.day
df['hour'] = datetime_ser.dt.hour
df['minute'] = datetime_ser.dt.minute
df['second'] = datetime_ser.dt.second
print(df)
df.info()
```

```python
#0 – 3.59 night, 4 – 6.59 early morning, 7 – 10.59 morning, 11 – 16.59 afternoon, 17 – 19.59 early evening, 20 – 23.59 evening
def daytime_category(time):
    if time >= 0 and time < 4:
        return "night"
    elif time < 7:
        return "early morning"
    elif time < 11:
        return "morning"
    elif time < 17:
        return "afternoon"
    elif time < 20:
        return "early evening"
    elif time < 24:
        return "evening"
    else:
        return "error"
    
#df['daytime'] = df['hour'].apply(daytime_category)
bins = [0,4,7,11,17,20,24]
labels = ['night', 'early morning', 'morning', 'afternoon', 'early evening', 'evening']
df['daytime'] = pd.cut(df['hour'], bins=bins, labels=labels, right=False)
print(df)
```

```python
df = df.set_index('user')
print(df)
```

```python
min_hour = df[df['daytime'] == 'morning']['hour'].min()
max_hour = df[df['daytime'] == 'night']['hour'].max()
print(f"Min: {min_hour}; Max = {max_hour}")

morning_visitors = df[(df['daytime'] == 'morning') & (df['hour'] == min_hour)]
print(f"Example: {morning_visitors[0]}")
hour_mode = df['hour'].mode()[0]
daytime_mode = df['daytime'].mode()[0]
print("Most common hour:", hour_mode)
print("Most common daytime:", daytime_mode)
```

```python
morning_visits = df[df['daytime'] == 'morning']

earliest_morning = morning_visits.nsmallest(3, ['hour', 'minute', 'second'])
print("3 earliest morning visits:")
print(earliest_morning[['hour', 'minute', 'second']])


latest_morning = morning_visits.nlargest(3, ['hour', 'minute', 'second'])
print("\n3 latest morning visits:")
print(latest_morning[['hour', 'minute', 'second']])
```

```python
stats = df.describe()
print("\nBasic statistics:")
print(stats)
```

```python
q1 = stats.loc['25%', 'hour']  # Первый квартиль (25-й процентиль)
q3 = stats.loc['75%', 'hour']  # Третий квартиль (75-й процентиль)
iqr = q3 - q1  # Межквартильный размах
print(f"\nInterquartile range (IQR) for hour: {iqr}")
print(f'Q1={q1}, Q3={q3}')
```
