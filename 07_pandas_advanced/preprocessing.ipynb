```python
import pandas as pd

df = pd.read_csv('../data/auto.csv', index_col='ID')
df
```

```python
initial_count = df.count()
print(f"Исходное количество:\n{initial_count}")
```

```python
df = df.drop_duplicates(subset=['CarNumber', 'Make_n_model', 'Fines'], keep='last')
print("Количество записей после удаления:", len(df))
df
```

```python
#NAN
print("До обработки:")
print(df.isnull().sum())
#----
df = df.dropna(thresh=len(df)-500, axis=1)
df
#----
df.loc[:, 'Refund'] = df['Refund'].ffill()
df.loc[:, 'Fines'] = df['Fines'].fillna(df['Fines'].mean())
df
```

```python
print("Пропуски после обработки:")
print(df.isnull().sum())
```

```python
df[['Make', 'Model']] = df['Make_n_model'].apply(
    lambda x: pd.Series(x.split(' ', 1)) if pd.notnull(x) else pd.Series([None, None])
)
df = df.drop('Make_n_model', axis=1)  # Удаляем исходный столбец
```

```python
df.to_json('../data/auto.json', orient='records')
```
