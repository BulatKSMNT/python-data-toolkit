```python
import pandas as pd
```

```python
df = pd.read_json('../data/auto.json').set_index('CarNumber')
df
```

```python
high_fines = df[df['Fines'] > 2100]
print(f"High fines:\n{high_fines}")
high_fines_refund = df[(df['Fines'] > 2100) & (df['Refund'] == 2)]
print(f"High fines:\n{high_fines_refund}")
focus_corolla = df[df['Model'].isin(['Focus', 'Corolla'])]
specific_cars = df.loc[['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS']]
```

```python
median_by_make = df.groupby('Make')['Fines'].median()
print(f"Median 1:\n{median_by_make}")
median_by_make_model = df.groupby(['Make', 'Model'])['Fines'].median()
print(f"Median 2:\n{median_by_make}")
count_by_make_model = df.groupby(['Make', 'Model'])['Fines'].count()
print(f"Count:\n{median_by_make}")
min_max_by_make_model = df.groupby(['Make', 'Model'])['Fines'].agg(['min', 'max'])
print(f"Minmax:\n{median_by_make}")
std_by_make_model = df.groupby(['Make', 'Model'])['Fines'].std()
display(std_by_make_model)
```

```python
freq_violators = df.groupby(level=0)['Fines'].count().sort_values(ascending=False)
print(freq_violators)
top1_violator = df.loc[[freq_violators.index[0]]]
print(top1_violator)
sum_fines_by_car = df.groupby(level=0)['Fines'].sum().sort_values(ascending=False)
print(sum_fines_by_car)
top1_sum_violator = df.loc[[sum_fines_by_car.index[0]]]
print(top1_sum_violator)
```

```python
cars_multiple_models = df.groupby(level=0)['Model'].nunique()
print(cars_multiple_models)
cars_with_conflicts = cars_multiple_models[cars_multiple_models > 1]
print(cars_with_conflicts)
```
