```python
import pandas as pd
import numpy as np
import requests

# Загружаем JSON-файл из предыдущего задания
df = pd.read_json('../data/auto.json')

# float с двумя знаками после запятой
pd.options.display.float_format = '{:.2f}'.format


print("Первые 5 строк данных:")
print(df.head())
print("\nИнформация о данных:")
print(df.info())
```

```python
np.random.seed(21)
sample = df.sample(n=200, random_state=21)

concat_rows = pd.concat([df, sample], ignore_index=True)

print(f"\nРазмер исходных данных: {df.shape}")
print(f"Размер объединенных данных: {concat_rows.shape}")
```

```python

np.random.seed(21)
years = pd.Series(np.random.randint(1980, 2020, size=len(concat_rows)), name='Year')

fines = pd.concat([concat_rows, years], axis=1)

print("\nДанные с добавленным годом выпуска:")
print(fines.head())
```

```python
# Загружаем список популярных фамилий
# path = "https://git.21-school.ru/students_repo/akikodru/DS_Bootcamp.Day05.ID_886517-1/-/blob/master/datasets/surname.json"
# response = requests.get(path) 
surnames = pd.read_json("../../datasets/surname.json")
surnames.columns = surnames.iloc[0]  
surnames = surnames.drop(0)  
print(surnames.columns)  

unique_cars = fines['CarNumber'].unique()
np.random.seed(21) 
owners = pd.DataFrame({
    'CarNumber': unique_cars,
    'SURNAME': np.random.choice(surnames['NAME'], size=len(unique_cars))
})


new_fines = pd.DataFrame({
    'CarNumber': ['D163P8161RUS', 'M247K5329RUS', 'T819R7045RUS', 'F356S1982RUS', 'L902W6437RUS'],
    'Make': ['Toyota', 'Ford', 'BMW', 'Audi', 'Mercedes'],
    'Model': ['Camry', 'Focus', 'X5', 'A4', 'C-Class'],
    'Fines': [1500.00, 2100.50, 3200.75, 1800.00, 2500.25],
    'Refund': [1, 2, 1, 2, 1],
    'Year': [2015, 2018, 2017, 2019, 2016]
})
fines = pd.concat([fines, new_fines], ignore_index=True)


owners = owners.iloc[:-20]
new_owners = pd.DataFrame({
    'CarNumber': ['B573Q2608RUS', 'G504D6239RUS', 'J837C1452RUS'],
    'SURNAME': ['IVAN', 'NIKITA', 'VLAD']
})
owners = pd.concat([owners, new_owners], ignore_index=True)

print("\nДатафрейм fines после добавления данных:")
print(fines.tail())
print("\nДатафрейм owners после изменений:")
print(owners.tail())
```

```python
# 1. Только номера, существующие в обоих датафреймах
inner_join = pd.merge(fines, owners, on='CarNumber', how='inner')

# 2. Все номера из обоих датафреймов
outer_join = pd.merge(fines, owners, on='CarNumber', how='outer')

# 3. Только номера из fines
left_join = pd.merge(fines, owners, on='CarNumber', how='left')

# 4. Только номера из owners
right_join = pd.merge(fines, owners, on='CarNumber', how='right')

print("\nРазмеры объединенных датафреймов:")
print(f"Inner join: {inner_join.shape}")
print(f"Outer join: {outer_join.shape}")
print(f"Left join: {left_join.shape}")
print(f"Right join: {right_join.shape}")
```

```python
# сводная таблицa
pivot_table = fines.pivot_table(
    values='Fines',
    index='Make',
    columns='Year',
    aggfunc='sum',
    fill_value=0
)

print("\nСводная таблица по штрафам:")
print(pivot_table)
```

```python

fines.to_csv('../data/fines.csv', index=False)
owners.to_csv('../data/owners.csv', index=False)

print("\nДанные успешно сохранены")
```
