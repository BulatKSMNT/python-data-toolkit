```python
import sqlite3
import pandas as pd
pd.set_option('display.width', 1000)
conn = sqlite3.connect('../data/checking-logs.sqlite') 
```

```python
schema = pd.io.sql.read_sql("PRAGMA table_info(checker);", conn)
print("Схема таблицы checker:")
print(schema)
```

```python
first_rows = pd.io.sql.read_sql("SELECT * FROM checker LIMIT 10;", conn)
print("\nПервые 10 строк таблицы checker:")
print(first_rows)
```

```python
query = """
SELECT COUNT(*) as count
FROM pageviews
WHERE uid IN (
    SELECT uid
    FROM checker
    WHERE status = 'ready'
    AND numTrials = 1
    AND labname IN ('laba04', 'laba04s', 'laba05', 'laba06', 'laba06s', 'project1')
)
"""

checkers = pd.io.sql.read_sql(query, conn)
print("\nРезультат подсчета:")
print(checkers)
```
