MovieLens Analytics Report: Исследование кинематографических предпочтений

Введение

Добро пожаловать в увлекательное путешествие по миру кинематографа! В этом отчете мы исследуем данные MovieLens - одной из самых популярных баз данных фильмов и рейтингов. Наша цель - раскрыть скрытые закономерности в предпочтениях зрителей, найти самые популярные жанры, проанализировать тренды по годам выпуска фильмов и многое другое.

Мы будем использовать первые 1000 записей из каждого файла датасета, что даст нам достаточно данных для выявления интересных паттернов, но при этом позволит работать эффективно.

1. Загрузка и инициализация данных

```python
%timeit
from movielens_analysis import Movies, Links, Ratings, Tags

movies = Movies('ml-latest-small/movies.csv')
links = Links('ml-latest-small/links.csv', 'ml-latest-small/movies.csv')
ratings = Ratings('ml-latest-small/ratings.csv', 'ml-latest-small/movies.csv')
tags = Tags('ml-latest-small/tags.csv')

print("Все данные успешно загружены!")
print(f"Загружено фильмов: {len(movies.movies)}")
print(f"Загружено рейтингов: {len(ratings.ratings)}")
print(f"Загружено тегов: {len(tags.tags)}")
```

2. Анализ фильмов по годам выпуска

Начнем наше исследование с изучения распределения фильмов по годам выпуска. Это поможет нам понять, какие периоды кинематографа наиболее представлены в нашей выборке.

```python
%timeit
release_distribution = movies.dist_by_release()

print("Топ-10 самых продуктивных лет в кинематографе:")
print("=" * 50)
for i, (year, count) in enumerate(list(release_distribution.items())[:10], 1):
    print(f"{i:2d}. {year} год: {count} фильмов")
```

Самым популярным периодом для кинематографа считаются 90-е

3. Жанровый ландшафт кинематографа

Теперь исследуем, какие жанры наиболее популярны среди фильмов в нашей базе данных.

```python
%timeit
genre_distribution = movies.dist_by_genres()

print("Топ-10 самых популярных жанров:")
print("=" * 40)
for i, (genre, count) in enumerate(list(genre_distribution.items())[:10], 1):
    print(f"{i:2d}. {genre:<15} {count:3d}")
```

Интересные факты:

Драма традиционно лидирует в кинематографе

Комедии и романы также очень популярны

Разнообразие жанров показывает богатство кинематографического мира

4. Фильмы с наибольшим количеством жанров

Некоторые фильмы сложно отнести к одному жанру. Давайте найдем самые "многожанровые" картины.

```python
%timeit
multi_genre_movies = movies.most_genres(10)

print("Топ-10 фильмов с наибольшим количеством жанров:")
print("=" * 60)
for i, (title, genre_count) in enumerate(multi_genre_movies.items(), 1):
    print(f"{i:2d}. {title:<40} ({genre_count} жанров)")
```

Анализ:

Многожанровые фильмы часто являются амбициозными проектами

Такие фильмы пытаются привлечь разную аудиторию

Это может как увеличить их популярность, так и размыть целевую аудиторию

5. Анализ рейтингов: взгляд зрителей

Перейдем к анализу того, как зрители оценивают фильмы. Сначала посмотрим на распределение оценок.

```python
%timeit
# Создаем объект для анализа фильмов через рейтинги
movie_ratings = ratings.Movies(ratings)

# Анализируем распределение оценок
rating_distribution = movie_ratings.dist_by_rating()

print("Распределение оценок зрителей:")
print("=" * 35)
for rating, count in rating_distribution.items():
    percentage = (count / sum(rating_distribution.values())) * 100
    print(f"{rating:.1f} ⭐ : {count:4d} ({percentage:5.1f}%)")
```

Психология оценок:

Большинство людей склонны давать положительные оценки

Крайне низкие оценки встречаются реже - люди просто не досматривают плохие фильмы

6. Самые обсуждаемые фильмы

Количество рейтингов может служить индикатором популярности и общественного интереса к фильму.

```python
%timeit
most_rated = movie_ratings.top_by_num_of_ratings(15)

print("Топ-15 самых обсуждаемых фильмов:")
print("=" * 50)
for i, (title, rating_count) in enumerate(most_rated.items(), 1):
    print(f"{i:2d}. {title:<35} ({rating_count} оценок)")
```

Феномен популярности:

Фильмы с большим количеством оценок часто являются культовыми

Это могут быть как шедевры, так и спорные картины

Количество обсуждений не всегда коррелирует с качеством

7. Лучшие фильмы по средним оценкам

Теперь посмотрим на фильмы с самыми высокими средними оценками.

```python
%timeit
top_average = movie_ratings.top_by_ratings(15, metric='average')

print("Топ-15 фильмов по средним оценкам:")
print("=" * 45)
for i, (title, avg_rating) in enumerate(top_average.items(), 1):
    print(f"{i:2d}. {title:<35} {avg_rating:.2f}")
```

8. Самые спорные фильмы

Некоторые фильмы вызывают полярные мнения. Найдем их по высокой дисперсии оценок.

```python
%timeit
controversial = movie_ratings.top_controversial(10)

print("Топ-10 самых спорных фильмов:")
print("=" * 40)
for i, (title, variance) in enumerate(controversial.items(), 1):
    print(f"{i:2d}. {title:<30} (дисперсия: {variance:.2f})")
```

Спорные шедевры:

Высокая дисперсия может указывать на новаторство

Такие фильмы часто становятся культовыми со временем

Они разделяют аудиторию на "за" и "против"

9. Анализ пользователей: кто и как оценивает

Теперь взглянем на самих пользователей и их поведение при оценке фильмов.

```python
%timeit
# Создаем объект для анализа пользователей
user_ratings = ratings.Users(ratings)

# Анализируем активность пользователей
user_activity = user_ratings.dist_by_num_of_ratings()

print("Статистика активности пользователей:")
print("=" * 40)
active_users = list(user_activity.items())[:10]
total_ratings = sum(user_activity.values())
total_users = len(user_activity)

print(f"Всего пользователей: {total_users}")
print(f"Всего оценок: {total_ratings}")
print(f"Среднее оценок на пользователя: {total_ratings/total_users:.1f}")
print("\nТоп-10 самых активных пользователей:")
for i, (user_id, rating_count) in enumerate(active_users, 1):
    print(f"{i:2d}. Пользователь {user_id}: {rating_count} оценок")
```

10. Самые строгие и добрые критики

Проанализируем пользователей по их средним оценкам.

```python
%timeit
user_averages = user_ratings.dist_by_avg_or_median(metric='average')

print("Самые добрые критики (высокие средние оценки):")
print("=" * 50)
kind_critics = list(user_averages.items())[-10:] 
for i, (user_id, avg_rating) in enumerate(reversed(kind_critics), 1):
    mood = "😊" if avg_rating > 4.0 else "🙂" if avg_rating > 3.5 else "😐"
    print(f"{i:2d}. Пользователь {user_id}: {avg_rating:.2f} {mood}")

print("\nСамые строгие критики (низкие средние оценки):")
print("=" * 50)
strict_critics = list(user_averages.items())[:10] 
for i, (user_id, avg_rating) in enumerate(strict_critics, 1):
    mood = "😤" if avg_rating < 2.0 else "😑" if avg_rating < 2.5 else "😑"
    print(f"{i:2d}. Пользователь {user_id}: {avg_rating:.2f} {mood}")
```

11. Самые противоречивые пользователи

Найдем пользователей, чьи оценки наиболее непоследовательны.

```python
%timeit
controversial_users = user_ratings.top_controversial_users(10)

print("Топ-10 самых непредсказуемых пользователей:")
print("=" * 50)
for i, (user_id, variance) in enumerate(controversial_users.items(), 1):
    print(f"{i:2d}. Пользователь {user_id}: дисперсия {variance:.2f}")
```

Типы зрителей:

Последовательные зрители дают стабильные оценки

Непредсказуемые могут быть более эмоциональными

Или у них очень разнообразные вкусы

12. Мир тегов: как зрители описывают фильмы

Теги дают уникальную возможность понять, как зрители воспринимают и описывают фильмы своими словами.

```python
%timeit
popular_tags = tags.most_popular(15)

print("Топ-15 самых популярных тегов:")
print("=" * 40)
for i, (tag, count) in enumerate(popular_tags.items(), 1):
    print(f"{i:2d}. '{tag}' - {count} использований")
```

13. Самые многословные теги

Некоторые пользователи любят детально описывать фильмы. Найдем самые длинные по количеству слов теги.

```python
%timeit
wordy_tags = tags.most_words(10)

print("Топ-10 самых многословных тегов:")
print("=" * 50)
for i, tag in enumerate(wordy_tags, 1):
    word_count = len(tag.split())
    print(f"{i:2d}. '{tag}' ({word_count} слов)")
```

14. Самые длинные теги

А теперь посмотрим на теги, которые длинные по количеству символов.


```python
%timeit
longest_tags = tags.longest(10)

print("Топ-10 самых длинных тегов:")
print("=" * 45)
for i, tag in enumerate(longest_tags, 1):
    char_count = len(tag)
    print(f"{i:2d}. '{tag}' ({char_count} символов)")
```

15. Теги-гибриды: длинные И многословные

Найдем теги, которые одновременно входят в топ по количеству слов и по длине.


```python
%timeit
hybrid_tags = tags.most_words_and_longest(10)

print("Теги-гибриды (одновременно многословные и длинные):")
print("=" * 60)
if hybrid_tags:
    for i, tag in enumerate(hybrid_tags, 1):
        word_count = len(tag.split())
        char_count = len(tag)
        print(f"{i:2d}. '{tag}'")
        print(f"{word_count} слов, {char_count} символов")
else:
    print("Интересно! Нет тегов, которые одновременно попадают в оба топа.")
    print("Это говорит о разнообразии стилей описания фильмов пользователями.")
```

16. Тематический поиск: теги об "экшене"

Давайте посмотрим, как пользователи описывают боевики и экшн-фильмы.

```python
%timeit
action_tags = tags.tags_with("action")

print("Теги, связанные с 'action':")
print("=" * 35)
if action_tags:
    for i, tag in enumerate(action_tags, 1):
        print(f"{i:2d}. '{tag}'")
    print(f"\nНайдено {len(action_tags)} тегов, связанных с экшеном")
else:
    print("Не найдено тегов со словом 'action'")
```

18. Самые Дорогие фильмы

```python
%%timeit -n1 -r1
try:
    expensive_movies = links.most_expensive(10)

    print("Топ-10 самых дорогих фильмов:")
    print("=" * 40)
    for i, (title, budget) in enumerate(expensive_movies.items(), 1):
        budget_millions = budget / 1_000_000
        print(f"{i:2d}. {title}")
        print(f"Бюджет: ${budget:,} ({budget_millions:.1f}M)")

except Exception as e:
    print("Не удалось получить данные о бюджетах фильмов")
    print(f"Ошибка: {str(e)}")
```
